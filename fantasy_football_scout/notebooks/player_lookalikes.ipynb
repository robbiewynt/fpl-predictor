```python
from difflib import SequenceMatcher
import pandas as pd

def similarity(a, b):
    return SequenceMatcher(None, a.lower(), b.lower()).ratio() * 100

def top_matches(index, left, right, top_n=10):
    name1, season1 = left.iloc[index][["full_name", "season"]]
    results = []

    for name2, season2 in right[["full_name", "season"]].itertuples(index=False):
        if season1 != season2:
            continue
        score = similarity(name1, name2)
        results.append({"name1": name1, "name2": name2, "season": season1, "score": score})

    results_df = pd.DataFrame(results)
    top_df = results_df.sort_values(by="score", ascending=False).head(top_n)
    return top_df

numbers = [88, 116, 135, 142, 149, 160, 167, 168, 194, 198, 202, 207, 233, 235, 258, 264, 276, 282, 327, 359, 387, 391, 395, 416, 445, 473, 484, 499, 505, 525, 547]

for number in numbers:
    print("---------------------------------------------------")
    print(top_matches(number, left, right))
    print("---------------------------------------------------")
```

```python
match_numbers = [1, 0, 1, 5, 5, 1, 1, 0, 1, 1, 1, 9, 1, 2, 3, 1, 2, 1, 1, 21, 2, 1, 2, 1, 1, 1, 0, 2, 1, 0, 2]

print(numbers[7])
print(numbers[19])
print(top_matches(168, left, right, 30))
print(top_matches(359, left, right, 30))
```

```python
for idx, match_pos in zip(numbers, match_numbers):
    # Get top matches for this index
    candidates = top_matches(idx, left, right, top_n=(match_pos + 1))
    
    # Pick the specific match position
    chosen = candidates.iloc[match_pos]
    
    # Find and update the original row instead of appending
    mask = (
        (matches_df['name1'] == chosen["name1"]) &
        (matches_df['season'] == chosen["season"])
    )

    if mask.any():
        matches_df.loc[mask, ['name2', 'score', 'match_indicator']] = [
            chosen["name2"], chosen["score"], 1
        ]
    else:
        # If not found (shouldn't happen normally), append as fallback
        matches_df = pd.concat([
            matches_df,
            pd.DataFrame([{
                "name1": chosen["name1"],
                "name2": chosen["name2"],
                "season": chosen["season"],
                "score": chosen["score"],
                "match_indicator": 1
            }])
        ], ignore_index=True)
```

```python
import random as rand

def find_lookalike(player_name, team, position, season, cost):
    # Filter players by team and position
    similar_players = player_data[
        (player_data['team'] == team) &
        (player_data['pos'] == position) &
        (player_data['season'] == season) &
        (player_data['full_name'] != player_name)
    ].copy()

    # Calculate cost difference as a new column
    similar_players['cost_diff'] = (similar_players['initial_cost'] - cost).abs()

    similar_players = similar_players.drop_duplicates(subset = ["full_name"])

    smallest_cost_diff = similar_players['cost_diff'].min()

    print(similar_players[similar_players['cost_diff'] == smallest_cost_diff])

    options = similar_players[similar_players['cost_diff'] == smallest_cost_diff]

    if len(options) == 1:
        print(options.iloc[0])
    else:
        selection = rand.randint(0, len(options) - 1)
        print(options.iloc[selection])


find_lookalike("Toti António Gomes", "WOL", "Defender", "24-25", 4.5)
```
